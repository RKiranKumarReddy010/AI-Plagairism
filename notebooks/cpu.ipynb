```python
import psutil

p = psutil.Process()          # current process
p.cpu_affinity([0, 1])        # pin to cores 0 and 1
print(p.cpu_affinity())       # verify

# Pin a different, already-running process by PID
other = psutil.Process(pid=15124)
other.cpu_affinity([2, 3])
```
